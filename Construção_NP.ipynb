```python
import ase
from ase.io import write
from ase.cluster import Cluster
from ase.cluster import wulff_construction
from ase.visualize import view
```

```python
surfaces = [(1, 0, 0), (1, 1, 0), (1, 1, 1)]
esurf = [1.0, 1.1, 0.9]   # Surface energies.
lc = 3.61000
size = 1300 # Number of atoms
atoms = wulff_construction('Cu', surfaces, esurf,
                           size, 'fcc',
                           rounding='closest', latticeconstant=lc)
```

```python
view(atoms, viewer='x3d')
```

```python
diameter = Cluster.get_diameter(atoms)/10
diameter
```

```python
write("Cu_cluster_3nm.xyz", atoms)
```
